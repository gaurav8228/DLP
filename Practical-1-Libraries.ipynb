**Aim:** Study Keras, PyTorch, TensorFlow and Theano.

This practical is mainly theoretical. Theano is included because it appears in the college manual; it is a legacy project and may not install on modern Python versions.

## 1. Keras
Keras is a high-level deep-learning API used to build and train neural networks with a relatively simple interface. In this repository we use `tf.keras`, the Keras API integrated with TensorFlow.

## 2. TensorFlow
TensorFlow is an open-source machine-learning framework. It represents data using tensors and provides tools for building, training and evaluating neural networks.

## 3. PyTorch
PyTorch is a deep-learning framework. Its `torch.nn` package provides neural-network layers, `torch.optim` provides optimizers, and autograd computes gradients.

## 4. Theano
Theano was a Python library for numerical computation and symbolic differentiation, with CPU/GPU support. It is now discontinued/legacy, so installation may fail on current Python versions. For this practical, study its purpose and features rather than relying on it for the later experiments.

```python
import tensorflow as tf
print("TensorFlow:", tf.__version__)

try:
    import torch
    print("PyTorch:", torch.__version__)
except ImportError:
    print("PyTorch is not installed in this environment.")

try:
    import keras
    print("Keras:", keras.__version__)
except ImportError:
    print("Keras is available through tf.keras:", tf.keras.__version__)
```

## Comparison

| Library | Main role | Key feature |
|---|---|---|
| Keras | High-level neural-network API | Simple model-building interface |
| TensorFlow | Machine-learning framework | Tensors, automatic differentiation, deployment tools |
| PyTorch | Deep-learning framework | Dynamic execution and autograd |
| Theano | Legacy numerical-computation library | Symbolic differentiation and graph optimization |
