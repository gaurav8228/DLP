# Practical 5 — Continuous Bag of Words (CBOW) using Word2Vec

**Aim:** Implement the CBOW model to predict a target word from surrounding context words.

**Dataset:** A small example corpus is provided in `dataset/corpus.txt`. You can replace it with a larger text corpus for better word embeddings.

```python
import re
from pathlib import Path
from gensim.models import Word2Vec

corpus_path = Path("dataset/corpus.txt")
if not corpus_path.exists():
    # Fallback for running the notebook from a different working directory
    corpus_path = Path("../dataset/corpus.txt")

text = corpus_path.read_text(encoding="utf-8").lower()
sentences = [
    re.findall(r"[a-z]+", line)
    for line in text.splitlines()
    if line.strip()
]
sentences = [s for s in sentences if s]
print("Number of sentences:", len(sentences))
print(sentences[:3])
```

## Train CBOW Word2Vec

In Gensim, `sg=0` selects CBOW (Continuous Bag of Words). `sg=1` selects Skip-gram.

```python
model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=2,
    min_count=1,
    workers=1,
    sg=0,
    epochs=100
)
print("Vocabulary:", list(model.wv.index_to_key))
```

```python
word = model.wv.index_to_key[0]
print("Example word:", word)
print("Vector shape:", model.wv[word].shape)
print("Vector:", model.wv[word])
```

```python
query_word = "learning"
if query_word in model.wv:
    print("Words similar to", query_word)
    print(model.wv.most_similar(query_word, topn=5))
else:
    print(query_word, "is not in this small corpus. Try one of:", model.wv.index_to_key[:10])
```

## Viva questions
1. What is Word Embedding? — Representing words as numerical vectors.
2. What is CBOW? — Predicting a target word using surrounding context words.
3. What is Skip-gram? — Predicting surrounding words from a target word.
4. What does `window=2` mean? — Use up to two words on each side as context.
5. What does `sg=0` mean in Gensim? — CBOW architecture.
6. What does `most_similar()` do? — Finds words with similar embedding vectors using cosine similarity.
