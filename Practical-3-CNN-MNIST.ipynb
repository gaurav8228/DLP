# Practical 3 — Image Classification using CNN

**Aim:** Implement a Convolutional Neural Network (CNN) for image classification using MNIST.

**Dataset:** MNIST is downloaded automatically by TensorFlow when first run.

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras import layers
print("TensorFlow:", tf.__version__)
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# CNN expects a channel dimension: (height, width, channels)
x_train = np.expand_dims(x_train, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)
print(x_train.shape, x_test.shape)
```

## Define CNN architecture

- Convolution: learns local image features such as edges and patterns.
- ReLU: introduces non-linearity.
- MaxPooling: reduces feature-map dimensions.
- Flatten: converts feature maps to a vector.
- Dense: performs final classification.

```python
model = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])
model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
history = model.fit(
    x_train, y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.1
)
```

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)
print("Test loss:", test_loss)
print("Test accuracy:", test_accuracy * 100, "%")
```

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Training")
plt.plot(history.history["val_accuracy"], label="Validation")
plt.title("CNN accuracy")
plt.xlabel("Epoch")
plt.legend()
plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Training")
plt.plot(history.history["val_loss"], label="Validation")
plt.title("CNN loss")
plt.xlabel("Epoch")
plt.legend()
plt.tight_layout()
plt.show()
```

## Viva questions
1. What is CNN? — A neural network especially suited to grid-like data such as images.
2. What is convolution? — Applying a filter/kernel over an image to extract features.
3. What is a feature map? — The output produced by a convolution filter.
4. What is pooling? — Down-sampling feature maps to reduce dimensions.
5. What is the role of Flatten? — Converts 2D/3D feature maps into a 1D vector.
6. Why CNN instead of only Dense layers? — CNNs exploit local spatial patterns and share filter weights.
