# Practical 2 — Feedforward Neural Network using Keras and TensorFlow

**Aim:** Implement a feedforward neural network on MNIST, train it using SGD, evaluate it and plot loss/accuracy.

**Dataset:** MNIST handwritten digits. TensorFlow downloads it on first use; internet is required unless the dataset is already cached locally.

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras import layers
print("TensorFlow:", tf.__version__)
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
print("Training images:", x_train.shape)
print("Testing images:", x_test.shape)
print("Training labels:", y_train.shape)
```

```python
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
print("Pixel range:", x_train.min(), "to", x_train.max())
```

## Build the feedforward network

`Flatten` converts each 28×28 image into 784 values. The hidden Dense layer has 128 neurons and ReLU activation. The output layer has 10 neurons, one for each digit (0–9).

```python
model = keras.Sequential([
    layers.Input(shape=(28, 28)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])
model.summary()
```

```python
model.compile(
    optimizer="SGD",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
print("Model compiled successfully")
```

```python
history = model.fit(
    x_train, y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)
```

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)
print("Test loss:", test_loss)
print("Test accuracy:", test_accuracy * 100, "%")
```

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Training accuracy")
plt.plot(history.history["val_accuracy"], label="Validation accuracy")
plt.title("Model accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Training loss")
plt.plot(history.history["val_loss"], label="Validation loss")
plt.title("Model loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
predictions = model.predict(x_test[:5], verbose=0)
for i in range(5):
    print("Predicted:", int(np.argmax(predictions[i])), "| Actual:", int(y_test[i]))
```

## Viva questions
1. What is a feedforward neural network? — Information moves from input through hidden layers to output; there are no feedback loops.
2. What is an epoch? — One complete pass through the training dataset.
3. What is batch size? — Number of training examples processed before a weight update.
4. Why normalize pixels? — To scale pixel values from 0–255 into 0–1.
5. What does ReLU do? — Returns max(0, x), adding non-linearity.
6. Why Softmax? — Converts output scores into class probabilities.
7. What is SGD? — Stochastic Gradient Descent, an optimization algorithm that updates weights using gradients.
